```python
# Графы
# Матрица смежности
def matrix_sm(n, edges):
    matrix = [[0]*n for _ in range(n)]

    for a, b in edges:
        matrix[a-1][b-1] = 1
        matrix[b-1][a-1] = 1

    return matrix


# Матрица инцидентности
def matrix_inc(n, edges):
    matrix = [[0]*len(edges) for _ in range(n)]

    for i in range(len(edges)):
        a, b = edges[i]
        matrix[a-1][i] = 1
        matrix[b-1][i] = 1

    return matrix


# Поиск компонент связности (DFS)
def find_components(n, edges):
    # список соседей
    adj = [[] for _ in range(n)]
    for a, b in edges:
        adj[a-1].append(b-1)
        adj[b-1].append(a-1)

    visited = [False]*n
    components = []

    for i in range(n):
        if not visited[i]:
            stack = [i]
            comp = []

            while stack:
                v = stack.pop()

                if not visited[v]:
                    visited[v] = True
                    comp.append(v+1)

                    for nei in adj[v]:
                        stack.append(nei)

            components.append(comp)

    return components
```

```python
# Узел дерева
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    # ВСТАВКА
    def insert(self, val):
        self.root = self._insert(self.root, val)

    def _insert(self, node, val):
        if node is None:
            return Node(val)

        if val < node.val:
            node.left = self._insert(node.left, val)
        elif val > node.val:
            node.right = self._insert(node.right, val)

        return node

    # ПОИСК
    def search(self, val):
      return self._search(self.root, val)

    def _search(self, node, val):
      if node is None or node.val == val:
        return node
      if val < node.val:
        return self._search(node.left, val)
      return self._search(node.right, val)

    # УДАЛЕНИЕ (упрощённое)
    def delete(self, val):
        self.root = self._delete(self.root, val)

    def _delete(self, node, val):
        if node is None:
            return None

        if val < node.val:
            node.left = self._delete(node.left, val)
        elif val > node.val:
            node.right = self._delete(node.right, val)
        else:
            # 1 или 0 детей
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left

            # 2 ребёнка → минимум справа
            min_node = node.right
            while min_node.left:
                min_node = min_node.left

            node.val = min_node.val
            node.right = self._delete(node.right, min_node.val)

        return node

# Пирамидальная сортировка
def heapify(arr, n, i):
    largest = i
    left = 2*i + 1
    right = 2*i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)


def heap_sort(arr):
    n = len(arr)

    # строим кучу
    for i in range(n//2 - 1, -1, -1):
        heapify(arr, n, i)

    # сортировка
    for i in range(n-1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)

    return arr
```
